Day 1 Overview:
1. Pull 5 years of WTI daily prices via yfinance (CL=F)
2. Compute daily log returns
3. Compute 30-day rolling annualized volatility
4. Find the date and value of the highest-volatility 30-day window

```python
import yfinance as yf
import numpy as np
import pandas as pd
```

```python
data = yf.download(tickers="CL=F", period='5y')
```

```python
print(data)
```

2. Compute daily log returns

```python
dailyLogReturns = np.log(data.Close/data.Close.shift(1))
dailyLogReturns = dailyLogReturns.rename(columns={"CL=F": "WTI Log Return"})
dailyLogReturns.columns.name = None
print(dailyLogReturns)
```

Annualized volatility = standard deviation of returns * sqrt(# trading days: 252)

```python
sd = dailyLogReturns.std()
tradingDays = 252
annualizedVol = sd*np.sqrt(tradingDays)
```

30-day rolling annualized volatility = standard deviation of a 30 day window * sqrt(252)

```python
rollingVol = dailyLogReturns.rolling(30).std()*np.sqrt(tradingDays)
rollingVol = rollingVol.rename(columns={"WTI Log Return": "30d Rolling Vol"})
print(rollingVol)
```

4. Find the date and value of the highest-volatility 30-day window

```python
maxDate = rollingVol['30d Rolling Vol'].idxmax()
maxVolValue = rollingVol['30d Rolling Vol'].loc[maxDate]
```

```python
print(f"Max vol date: {maxDate.date()}")
print(f"Max vol value: {maxVolValue:.1%}", )
```
