```python
import pandas as pd
import numpy as np
from core.returns import log_returns
```

**Instructions**
1. Create a DataFrame with three columns: 'date', 'ticker', 'close'
   — populate it with at least 5 rows of made-up data for two tickers (e.g., 'CL' and 'NG')
   — use date strings like '2026-05-01'

2. Set the 'date' column as the index

3. Select all rows for one ticker using boolean filtering

4. Calculate the mean close price for each ticker using groupby

```python
date = pd.Series(['2026-05-01','2026-05-01','2026-05-02','2026-05-03','2026-05-02'])
ticker = pd.Series(['CL','NG','CL','CL','NG'])
close = pd.Series([100,88,101,102,89])

df = pd.DataFrame({'date': date, 'ticker': ticker, 'close': close})
print(df)
```

2. Set the 'date' column as the index

```python
df = df.set_index('date')
print(df)
```

3. Select all rows for one ticker using boolean filtering

```python
print(df[df['ticker'] == 'CL'])
```

4. Calculate the mean close price for each ticker using groupby

```python
print(df.groupby(by='ticker')['close'].mean())
```

5. Convert date strings to datetime

```python
df.index = pd.to_datetime(df.index)
df = df.sort_index()
print(df.index.dtype)
print(df)

sliced_df = df.loc['2026-05-01':'2026-05-02']
print(sliced_df)
```

6. Introduce missing data: Manually set one close value to np.nan. Then: detect which rows have missing data using isna(), drop them with dropna(), and separately show fillna() using the forward-fill method. This is bread-and-butter data cleaning.

```python
mask = (df.index == '2026-05-02') & (df['ticker'] == 'CL')
print(df.loc[mask, 'close'])
df.loc[mask, 'close'] = None
print(df.loc[mask, 'close'])
print(df)
print(df.isna())
df_drop_na = df.dropna()
print(df_drop_na)
```

6b. ffill NaN

```python
print(df)
df['close'] = df.groupby(by='ticker')['close'].ffill()
print(df)
```

7. Connect to your library: Import log_returns from core.returns and calculate log returns on just the CL close prices. This gets you practicing the full pipeline: filter → extract a Series → pass to your library function.

```python
cl_mask = (df['ticker'] == 'CL')
cl_close = df.loc[cl_mask, 'close']
print(cl_close)

l_returns = log_returns(cl_close)
print(f"Log Returns: {l_returns: 0.4f}")
```
